### Data Cleaning

```python
# Create a mapping dictionary
replace_map = {
    "VirginiaAmerican": "Us",
    "Minnesota46826/N/A9": "Minnesota",
    "Tajikistan\tMilitary": "Tajikistan",
    "Spain\r\n\t\r\nMoron": "Spain",
    "Indonesia\r\n\t\r\nSarmi": "Indonesia",
    "Djbouti\r\n\tDjbouti": "Djbouti",
    "Azerbaijan\r\n\t\r\nBakou": "Azerbaijan",
    "Bosnia-Herzegovina": "Bosnia",
    "Brazil\tAmazonaves": "Brazil",
    "Brazil\tLoide": "Brazil",
    "Brazil\r\nFlorianopolis": "Brazil",
    "Chile\tAerolineas": "Chile",
    "DCCapital": "Washington",
    "Norway\tCHC": "Norway",
    "India\tPawan": "India",
    "NWT Canada": "Canada",
    "UAEGulf": "UAE",
    "Covington/Hebron": "Covington",
    "Airlines": "",
    "Air": "",
    "''"
    "NWT Canada": "Canada",
    "UKBritish": "UK",
    "Norway\tChc": "Norway",
    "Virgin": "Virginia"
}

# Apply replacements
df['Country/Region'] = df['Country/Region'].replace(replace_map)

# Confirm changes
print(df['Country/Region'].unique())
```

```python
# Correct the misspelled manufacturer names
df['Aircraft Manufacturer'] = df['Aircraft Manufacturer'].replace(
    {'Dormier': 'Dornier', 'Donier': 'Dornier', 'Domier': 'Dornier'}
)
```

```python
# Replace incorrect manufacturer names
df['Aircraft Manufacturer'] = df['Aircraft Manufacturer'].replace({
    'Aerospeciale': 'Aerospatiale',
    'Aerospetiale': 'Aerospatiale',
    'AérospatialeAeritalia': 'Aerospatiale'
})
```

```python


df['Aircraft Manufacturer'] = df['Aircraft Manufacturer'].replace({
    'Swallow_X000D_\\nSwallow': 'Swallow',
    'Swallow\\r\\nSwallow': 'Swallow'
})
```

```python
df['Aircraft Manufacturer'] = df['Aircraft Manufacturer'].replace({
    'Ilushin': 'Illyushin',
    'Iluyshin': 'Illyushin',
    'Ilyushin': 'Illyushin',
    'Ilysushin': 'Illyushin',
})
```

```python
df['Aircraft Manufacturer'] = df['Aircraft Manufacturer'].replace({
    'Lockhed': 'Lockheed',
    'Unknown': 'N/A',
    'Mcdonnel': 'Mcdonnell',
    'Lioré': 'Liore'
})
```

```python
df['Aircraft Manufacturer'] = df['Aircraft Manufacturer'].replace({
    'Shortss': 'Short',
    'Tuolev': 'Tupolev'
})
```

```python
# Replace 's' with 'Phillipines' in 'Country/Region'column
df['Country/Region'] = df['Country/Region'].replace('s', 'Phillipines')
```

```python
# replace 'Zepplin' with 'Zeppelin' in all columns
df = df.replace(r'\bZepplin\b', 'Zeppelin', regex=True)
```

```python
# change column values to propercase
df['Country/Region'] = df['Country/Region'].str.title()
df['Aircraft Manufacturer'] = df['Aircraft Manufacturer'].str.title()
df['Aircraft'] = df['Aircraft'].str.title()
df['Location'] = df['Location'].str.title()
df['Operator'] = df['Operator'].str.title()
```

```python
df = df.fillna('N/A')
```
